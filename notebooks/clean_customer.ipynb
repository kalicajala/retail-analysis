```python
# import libraries
import numpy as np
import pandas as pd
```

```python
# read in the csv file
df = pd.read_csv('../data/raw/customer_data.csv')
print(df.head)
```

```python
# get count of missing values per column
print(df.isnull().sum())
```

```python
# email is only column with missing values
# column contents aren't relevant to our analysis
# drop the column
df.drop('email', axis=1, inplace=True)
print(df.columns.tolist())
```

```python
# 2 unnamed columns appear
# only going to keep customer id, age, gender, and city
df = df[['customer_id', 'age', 'gender', 'city']]

# check remaining columns
print(df.columns.tolist())
```

```python
# check for duplicates
print(df[df.duplicated()])
```

```python
# customer_id
# check for correct data type
df['customer_id'].dtype

# check for consistent customer ID format
df['customer_id'].str.len().value_counts()
```

```python
# age
# check that age is numeric
df['age'].dtype

# check that age ranges are realistic
df['age'].describe()
```

```python
# gender
# check that there is only Male, Female, and Other
df['gender'].value_counts()

# replace ??? with new 'Unknown' category
df['gender'] = df['gender'].replace('???', 'Unknown')
df['gender'].value_counts()
```

```python
# city
# check the values in this column
df['city'].value_counts()
```

```python
# write the cleaned data to a new csv file
df.to_csv('../data/cleaned/customer_data_cleaned.csv')
```
