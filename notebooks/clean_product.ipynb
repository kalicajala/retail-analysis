```python
# import libraries
import numpy as np
import pandas as pd
```

```python
# read in the csv file
df = pd.read_csv('../data/raw/product_data.csv')
print(df.head)
```

```python
# check for null values in each column
print(df.isnull().sum())
```

```python
# replace null values in color column with "Unknown"
df['color'] = df['color'].fillna("Unknown")
# check for null values
print(df.isnull().sum())
```

```python
# category
# check category values
df['category'].value_counts()
```

```python
# replace '???' with 'Unknown'
df['category'] = df['category'].replace('???', 'Unknown')
df['category'].value_counts()
```

```python
# color
# check category values
df['color'].value_counts()
```

```python
# size
# check category values
df['size'].value_counts()
```
